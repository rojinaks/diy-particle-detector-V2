# Alter diy detektor signal/noise analyse


```python
import pandas as pd
import numpy as np
import os

# Pfad zum Ordner
folder = "/Users/rojinaksu/PycharmProjects/diy-particle-detector-V2/codes/Signal tests/messungen_BA"

# ---------------------
# Funktion zur Leistungsberechnung für Datei-Bereich
def berechne_leistungen(start, ende):
    files = [os.path.join(folder, f"Tek{str(i).zfill(3)}.csv") for i in range(start, ende + 1)]
    leistungen = []

    for fname in files:
        try:
            df = pd.read_csv(fname, skiprows=8)
            voltage = pd.to_numeric(df.iloc[:, 1], errors='coerce')

            # Entferne NaN, inf, -inf
            voltage = voltage[np.isfinite(voltage)]
            voltage = voltage[np.abs(voltage) < 1e3]

            if len(voltage) == 0:
                print(f"Nur ungültige Werte in Datei: {fname}")
                continue

            power = np.mean(voltage**2)
            leistungen.append(power)

        except Exception as e:
            print(f"Fehler bei Datei {fname}: {e}")

    return leistungen

# ---------------------
# 1. Rauschleistung berechnen
noise_powers = berechne_leistungen(141, 192)
mean_noise_power = np.mean(noise_powers)
print(f"\n{len(noise_powers)} Rauschdateien verarbeitet.")
print(f"Mittlere Rauschleistung (V²): {mean_noise_power}")

# ---------------------
# 2. Signalamplitude aus Tek025.csv
signal_file = os.path.join(folder, "Tek325.csv")
try:
    df_signal = pd.read_csv(signal_file, skiprows=8)
    voltage_signal = pd.to_numeric(df_signal.iloc[:, 1], errors='coerce')
    voltage_signal = voltage_signal[np.isfinite(voltage_signal)]
    voltage_signal = voltage_signal[np.abs(voltage_signal) < 1e3]

    if len(voltage_signal) > 0:
        amplitude = np.abs(np.min(voltage_signal))  # negativer Peak
        print(f"\nSignalamplitude aus Tek025.csv: {amplitude:.3f} V")
    else:
        print("\nKeine gültigen Werte in Tek025.csv")
except Exception as e:
    print(f"\nFehler beim Laden von Tek025.csv: {e}")

```

```python
import math

noise_rms = math.sqrt(mean_noise_power)
snr_amp_db = 20 * math.log10(amplitude / noise_rms)
print(f"\nAmplitude-basiertes SNR: {snr_amp_db:.2f} dB")
```
