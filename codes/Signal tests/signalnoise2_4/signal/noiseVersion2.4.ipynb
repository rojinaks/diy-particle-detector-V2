```python
import pandas as pd
import numpy as np
import os
import math

folder = "/Users/rojinaksu/PycharmProjects/diy-particle-detector-V2/codes/Signal tests/messungen2_4"

# ---------------------
# Funktion zur Leistungsberechnung (RMS²)
def berechne_leistungen(start, ende):
    files = [os.path.join(folder, f"Tek{str(i).zfill(3)}.csv") for i in range(start, ende + 1)]
    leistungen = []

    for fname in files:
        try:
            df = pd.read_csv(fname, skiprows=8)
            voltage = pd.to_numeric(df.iloc[:, 1], errors='coerce')

            voltage = voltage[np.isfinite(voltage)]
            voltage = voltage[np.abs(voltage) < 1e3]

            if len(voltage) == 0:
                print(f"Nur ungültige Werte in Datei: {fname}")
                continue

            power = np.mean(voltage**2)  # RMS²
            leistungen.append(power)

        except Exception as e:
            print(f"Fehler bei Datei {fname}: {e}")

    return leistungen

# ---------------------
# Funktion zur Amplitudenberechnung
def berechne_amplituden(start, ende):
    files = [os.path.join(folder, f"Tek{str(i).zfill(3)}.csv") for i in range(start, ende + 1)]
    amplituden = []

    for fname in files:
        try:
            df = pd.read_csv(fname, skiprows=8)
            voltage = pd.to_numeric(df.iloc[:, 1], errors='coerce')

            voltage = voltage[np.isfinite(voltage)]
            voltage = voltage[np.abs(voltage) < 1e3]

            if len(voltage) == 0:
                print(f"Nur ungültige Werte in Datei: {fname}")
                continue

            amplitude = np.abs(np.min(voltage))  # negativer Peak
            amplituden.append(amplitude)

        except Exception as e:
            print(f"Fehler bei Datei {fname}: {e}")

    return amplituden

# ---------------------
# 1. Noise RMS²
noise_powers = berechne_leistungen(53, 63)
mean_noise_power = np.mean(noise_powers)
noise_rms = math.sqrt(mean_noise_power)
print(f"\n{len(noise_powers)} Rauschdateien verarbeitet.")
print(f"Mittlere Rauschleistung (V²): {mean_noise_power}")
print(f"Mittleres Rauschen RMS (V): {noise_rms}")

# ---------------------
# 2. Mittlere Signalamplitude
signal_amplituden = berechne_amplituden(0, 52)
mean_amplitude = np.mean(signal_amplituden)
print(f"\n{len(signal_amplituden)} Signaldaten verarbeitet.")
print(f"Mittlere Signalamplitude (V): {mean_amplitude}")

# ---------------------
# 3. Amplitudenbasiertes SNR in dB (Gleichung 4.3)
snr_amp_db = 20 * math.log10(mean_amplitude / noise_rms)
print(f"\nAmplitude-basiertes SNR: {snr_amp_db:.2f} dB")
```
