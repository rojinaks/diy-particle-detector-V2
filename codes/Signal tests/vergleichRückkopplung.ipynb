```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Pfad zum Ordner
folder = "/Users/rojinaksu/PycharmProjects/diy-particle-detector-V2/codes/Signal tests/messungen_BA"

# Datei-Indizes
file_indices = list(range(21, 26)) + [27]

# Plot vorbereiten
plt.figure(figsize=(10, 5))

for i in file_indices:
    fname = os.path.join(folder, f"Tek{str(i).zfill(3)}.csv")
    try:
        df = pd.read_csv(fname, skiprows=8)
        time = pd.to_numeric(df.iloc[:, 0], errors='coerce')
        voltage = pd.to_numeric(df.iloc[:, 1], errors='coerce')

        plt.plot(time, voltage, label=f"Tek{str(i).zfill(3)}")
    except Exception as e:
        print(f"Fehler bei Datei {fname}: {e}")

# Plot-Details
plt.xlabel("Zeit (s)")
plt.ylabel("Spannung (V)")
plt.title("Spannungsverläufe der Dateien Tek021–025 und Tek027")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
